```python
import matplotlib.pyplot as plt
import random

# 设置折线图支持中文
plt.rcParams['font.sans-serif'] = ['SimHei']

beijing_x = [i for i in range(1,25)]
beijing_y = [random.randint(10,15) for i in range(24)]

xian_x = [i for i in range(1,25)]
xian_y = [random.randint(12,18) for i in range(24)]



# 修改上述代码

# 设置画布大小
plt.figure(figsize=(12,8), dpi=100)

# 设置标题  fontsize = 18 : 标题字体大小，pad = 15 : 标题距离下方15
plt.title("气温24小时变化曲线", fontsize = 18, pad = 15)


plt.plot(beijing_x,beijing_y, color = "blue", label = '北京')
plt.plot(xian_x,xian_y, color = "red", label = '西安')                    
                            
# 设置x y 的标签
plt.xlabel("时间", fontsize = 14, labelpad = 10)
plt.ylabel("温度℃", fontsize = 14, labelpad = 10)

# 设置 x刻度
plt.xticks(beijing_x)

# 设置 y刻度

ytick_values = [i for i in range(5,21)]
plt.yticks(ytick_values)

# 设置网格线    visible=True : 显示网格线，linestyle='--' : 网格线为虚线，linewidth=0.5 : 网格线宽度为0.5，alpha=0.5 : 网格线透明度为0.5
plt.grid(visible=True, linestyle='-.', linewidth=0.5 , alpha=0.5)
# 显示图例
plt.legend(loc = "upper right", fontsize = 12)

plt.show()
```
