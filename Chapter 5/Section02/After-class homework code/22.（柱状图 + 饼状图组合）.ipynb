```python
"""
    某奶茶店统计了 4 款饮品的月销量及顾客支付方式占比：

    - 饮品月销量：珍珠奶茶 520 杯、杨枝甘露 468 杯、烤奶 312 杯、柠檬茶 275 杯
    - 支付方式占比：微信支付 55%、支付宝 35%、现金 10%
    
    请编写代码：
    
    1. 创建 1 行 2 列的子图，画布大小为 (16, 6)
    2. 左侧子图绘制 4 款饮品销量的柱状图
    3. 右侧子图绘制支付方式占比的饼状图，要求显示百分比（保留 1 位小数）并带上类别标签
    4. 为两个子图分别设置标题

"""
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']
# 绘制一行两列的画布
figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(16,6))

ax1 = axes[0]
ax2 = axes[1]

# 左侧子图：饮品月销量柱状图
drinks = ['珍珠奶茶', '杨枝甘露', '烤奶', '柠檬茶']
sales = [520, 468, 312, 275]

# 绘制柱状图
ax1.bar(drinks,sales, color='pink', width=0.25)

ax1.grid(True, linestyle='--', linewidth=0.5, alpha=0.5)

# 设置柱状图的标题和轴标签
ax1.set_title("饮品月销量")
ax1.set_xlabel("饮品")
ax1.set_ylabel("销量")

# 右侧子图：支付方式占比饼状图
payment_methods = ['微信支付', '支付宝', '现金']
payment_method_proportions = [0.55, 0.35, 0.10]
# 绘制饼状图
ax2.pie(payment_method_proportions, labels=payment_methods, autopct="%1.1f%%")
# 设置饼状图的标题和图例
ax2.set_title("支付方式占比")
ax2.legend(loc="lower center", ncol=3, fontsize=10, bbox_to_anchor=(0.5, -0.1))
```
