```python
"""
    某环境监测站记录了某市一周的空气质量指数（AQI）：
    
      星期  	周一  	周二  	周三  	周四  	周五  	周六  	周日  
      AQI 	105 	132 	98  	150 	87  	76  	93  
    
    请编写完整的 Python 代码：
    
    1. 导入 Matplotlib 并解决中文显示问题
    2. 绘制 AQI 随星期变化的折线图（折线颜色为绿色）
    3. 设置标题为"某市一周空气质量变化"、X 轴标签为"星期"、Y 轴标签为"AQI"
    4. 显示网格线并展示图形

"""

import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']  # 支持中文显示

aqi = [105, 132, 98, 150, 87, 76, 93]
week = ["周一", "周二", "周三", "周四", "周五", "周六", "周日"]

# 绘制折线图
plt.plot(week, aqi, color="green")

# 设置折线图的标题
plt.title("某市一周空气质量变化")

# 
plt.grid(True, linestyle="--", linewidth=0.5, alpha=0.5)

# 设置 x轴的标签
plt.xlabel("星期")
# 设置 y轴的标签
plt.ylabel("AQI")

plt.show()
```
