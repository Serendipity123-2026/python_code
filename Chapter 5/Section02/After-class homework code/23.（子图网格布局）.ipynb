```python
"""
    某健身 App 记录了会员数据：

    - 近 6 个月新增会员数：1月 320、2月 280、3月 410、4月 390、5月 450、6月 505
    - 会员年龄构成：18-25 岁 40%、26-35 岁 35%、36-45 岁 15%、46 岁及以上 10%
    
    请编写代码：
    
    1. 创建 2 行 2 列的子图网格，并设置整个画布的总标题"会员数据分析"
    2. 在第一行第二列的子图中绘制近 6 个月新增会员数的折线图（含标题）
    3. 在第二行第一列的子图中绘制会员年龄构成的饼状图（含标题和百分比）
    4. 其余两个子图留空不画

"""

import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei'] 

figure, axes = plt.subplots(nrows=2,ncols=2,figsize=(25,20))

ax1 = axes[0,0]
ax2 = axes[0,1]
ax3 = axes[1,0]
ax4 = axes[1,1]

# 画布的标题
plt.suptitle("会员数据分析", fontsize=20)


# 获取数据
month = ['1月', '2月', '3月', '4月', '5月', '6月']
member_count = [320, 280, 410, 390, 450, 505]

# 绘制折线图
ax2.plot(month,member_count)
# 设置折线图标题
ax2.set_title("不同月份新增会员数量", fontsize=16, pad=20)
# 设置网格
ax2.grid(True, linestyle="--",linewidth=0.5, alpha=0.5)
# xy轴标签
ax2.set_xlabel("月份", fontsize=16)
ax2.set_ylabel("会员数量", fontsize=16)


# 获取饼状图数据
age_groups = ['18-25岁', '26-35岁', '36-45岁', '46岁及以上']
proportions = [0.40, 0.35, 0.15, 0.10]
# 绘制饼状图
ax3.pie(proportions, labels=age_groups, autopct="%1.1f%%")
ax3.set_title("会员年龄构成", fontsize=16, pad=20)
ax3.legend(loc="lower center", ncol=2, fontsize=10, bbox_to_anchor=(0.5, -0.01))
plt.show()
```
