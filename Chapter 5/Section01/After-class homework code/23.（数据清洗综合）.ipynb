```python
"""
    某门店营业额 DataFrame df 存在以下数据质量问题：

    营业额 列有部分负数（录入错误）
    营业日期 列中日期分隔符不统一，有的用 / 有的用 -
    存在少量缺失值和完全重复的记录

    请写出完整的清洗代码：修正负数营业额、统一日期分隔符为 -、删除缺失值所在行、删除重复记录

"""

import pandas as pd

df = pd.read_csv("./resources/store_revenue.csv")

# 对营业额中负数部分取绝对值
df["营业额"] = df["营业额"].abs()

# 统一营业时间格式
df['营业日期'] = df['营业日期'].str.replace("/", "-")

# 删除缺失值
df.dropna(inplace=True)

# 查看重复的营业时间数据
# df[df.duplicated("营业日期")]

# 删除重复的营业时间数据
df.drop_duplicates(subset="营业日期", inplace=True)
df
```
