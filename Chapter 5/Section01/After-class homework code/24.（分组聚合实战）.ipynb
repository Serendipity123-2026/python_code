```python
"""
    某电影票订票平台 DataFrame df 包含列：影院名、影片、票价、张数。

    请编写代码完成：
    
    1. 新增一列 消费金额 = 票价 × 张数
    2. 按影院分组，统计每个影院的总消费金额
    3. 按影院分组，一次统计出每个影院的订单数、总消费金额、平均票价（用 agg 实现）
    4. 将第 2 步的结果保存到 cinema_revenue.csv，要求不带行索引

"""

import pandas as pd

df = pd.read_csv("./resources/cinema_orders.csv")

# 新增消费金额列
df['消费金额'] = df['票价'] * df['张数']

# 按影院分组，统计每个影院的总消费金额
result = df.groupby('影院名')['消费金额'].sum()
print(result)

result.to_csv("cinema_revenue.csv", index=False)

# 按影院分组，一次统计出每个影院的订单数、总消费金额、平均票价
df.groupby('影院名').agg({
    '影片': "count",
    '消费金额': "sum",
    '票价': "mean"
}).round(2)
```
